```python
import lightkurve as lk
import numpy as np
import gc # Import the garbage collection module

# Search and download all light curves for TIC 25155310
# Using author='SPOC' to retrieve data processed by the Science Processing Operations Center pipeline
search_result = lk.search_lightcurve('TIC 25155310', author='SPOC')
lc = search_result.download_all()

# Stitch the light curves from different sectors together
lc = lc.stitch()

# Optionally, downsample the light curve to reduce its size before further processing
# This is a common strategy to mitigate memory issues.
# Adjust binsize as needed - a larger binsize means more aggressive downsampling.
# For example, binsize=10 reduces the number of data points by a factor of 10.
lc_binned = lc.bin(binsize=30)

# Explicitly trigger garbage collection to free up memory before the potentially memory-intensive outlier removal step
# This can help release memory held by objects that are no longer needed.
gc.collect()

# Remove outliers using sigma clipping from the binned light curve
lc_cleaned = lc_binned.remove_outliers(sigma=5)

# Normalize the light curve for easier comparison and analysis
lc_normalized = lc_cleaned.normalize()

# Plot the raw (stitched), binned, and cleaned light curves
ax = lc.plot(label='Original Light Curve (Stitched)')
lc_binned.plot(ax=ax, label=f'Binned Light Curve (binsize={30})')
lc_normalized.plot(ax=ax, label='Cleaned & Normalized Light Curve')

# Flatten the light curve to remove long-term stellar variability or instrumental drift
# window_length determines the size of the filter window, polyorder determines the polynomial order
lc_flattened = lc_normalized.flatten(window_length=1501, polyorder=3)

# Plot the normalized and flattened light curves
ax = lc_normalized.plot(label='Normalized Light Curve')
lc_flattened.plot(ax=ax, label='Flattened Light Curve')
```
